# Imported libraries

```python
import pandas as pd
import numpy as np
import random
```

# --> Data Creation ( 1000 sales record )

```python
cities =[
    "Lahore",
    "Karachi",
    "Islamabad",
    "Faisalabad",
    "Multan",
    "Peshawar",
    "Rawalpindi"
]
```

```python
categories=[
    "Electronics",
    "Grocery",
    "Clothing",
    "Sports",
    "Furniture"
]
```

```python
products ={
    "Electronics" :["Mobile","Laptop","Headphones","SmartWatch"],
    "Clothing" : ["T-shirts","Jacket", "Shoes","Pant"],
    "Grocery": ["Rice", "Milk", "Bread", "Eggs"],
    "Sports" : ["Bat" , "Football" , "Tennis Ball" ,"Gloves" , "Wickets" , "Basket Ball"],
    "Furniture" : ["Chair" , "Table" , "Sofa" , "Bed" , "Desk" ]
}
```

```python
first_name = [
    "Ali" , "Zain" , "Sara" , "Javaid" , "Ahmed" , "Tanveer" , "Eman" , "Usman" , "Waleed" , "John" , "James" , "Peter" , "Jashaid" , "Taha" , "Zainab" ,"Iqra" , "Zulaiha"
]

last_name = [
    "Ahmed" , "Nasir" , "Tariq" , "Azhar" , "Sohail" , "Junaid" , "Nisar" , "Qamar" , "Malik" , "Qasim" , "Moiz" , "Hashir" , "Imran" , "Shahab" , "Kamran" , "Khan" , "Usmani"
            ]
```

```python
payment_method = [
    "Cash",
    "Card",
    "Online"
]
```

```python
returned = [
    "Yes",
    "No"
]
```

```python
sales_data = []
```

```python
category= random.choice(categories)

product = random.choice(products[category])

quantity = random.randint(1,5)

unit_price = random.randint(500 , 50000)

discount = random.choice([0 , 5 , 10 , 15 , 20])

gross_sale = quantity * unit_price
total_sale = gross_sale - ( gross_sale * discount / 100 )
```

```python
record = {
    "OrderID" : 1,
    "Name" : random.choice(first_name) +" "+ random.choice(last_name),
    "City" : random.choice(cities),
    "Category" : category,
    "Product" : product,
    "Quantity" : quantity,
    "Unit Pice" : unit_price,
    "Discount" : discount,
    "Total Sales": total_sale,
    "Payment Method" : random.choice(payment_method),
    "Returned" : random.choice(returned)
    
}
```

```python
print(record)
```

```python
for i in range(1000):
    category= random.choice(categories)
    product = random.choice(products[category])
    quantity = random.randint(1,5)
    unit_price = random.randint(500 , 50000)
    discount = random.choice([0 , 5 , 10 , 15 , 20])
    gross_sale = quantity * unit_price
    total_sale = gross_sale - ( gross_sale * discount / 100 )

    record = {
    "OrderID" : 1 + 1,
    "Name" : random.choice(first_name) +" "+ random.choice(last_name),
    "City" : random.choice(cities),
    "Category" : category,
    "Product" : product,
    "Quantity" : quantity,
    "Unit Pice" : unit_price,
    "Discount" : discount,
    "Total Sales": total_sale,
    "Payment Method" : random.choice(payment_method),
    "Returned" : random.choice(returned)
    }
    sales_data.append(record)
    


    
```

# --> Convert the data into Data Frame

```python
df = pd.DataFrame(sales_data)
```

```python
print(df)
```

# Adding DATE column into the Dataset

```python
from datetime import datetime , timedelta
```

```python
start_date = datetime(2026 , 1 , 1)
end_date = datetime(2026 , 12 , 31)

dates_list = []
for i in range(len(df)):
    random_days = random.randint( 0 , (end_date - start_date).days)
    date = start_date + timedelta(days = random_days)
    dates_list.append(date)
```

```python
df["Dates"] = dates_list
```

# 3. Which category generated the most revenue

```python
category_sales = (df.groupby("Category")["Total Sales"].sum())

best_category = category_sales.idxmax()
best_revenue = category_sales.max()

print("Category :",best_category)
print("Revenue :",best_revenue)
```

# 4. Which Payment method has been used most by the customer

```python
pay_counts = df.groupby("Payment Method").size()

most_used = pay_counts.idxmax()
count = pay_counts.max()

print(f"Most used payment method : {most_used}") 
print(f"Number of time used : {count}")
```

# 5. Get the Average Order Value

```python
average_order_value = df["Total Sales"].mean()

print(average_order_value)
```

# 6. Which customer has highest average order money

```python
average_money = df.groupby("Name")["Total Sales"].mean()
most_spender = average_money.idxmax()
money = average_money.max()

print(f"Highest average order money spent by : {most_spender}")
print(f"Money spent : {money}")
```

# 11. Which City generated the highest Electronics revenue 

```python
electronics = df[df["Category"]=="Electronics"]
elec_rev_city = electronics.groupby("City")["Total Sales"].sum()

highest_rev_city = elec_rev_city.idxmax()
highest_rev = elec_rev_city.max()

print(f"Highest revenue city : {highest_rev_city}")
print(f"Highest revenue : {highest_rev}")
```

# 12. How much revenue did Lahore generate from Electronics?

```python
lahore_elec_rev =df[ (df["City"]=="Lahore") & (df["Category"]=="Electronics")]["Total Sales"].sum()
print(f"Total Revenue Lahore generated from elctrronics category products is : {lahore_elec_rev}")
```

# --> Data Visualization

```python
import matplotlib.pyplot as plt
```

# BAR chart (Top 5 poducts by revenue)

```python
Top_5_products = df.groupby("Product")["Total Sales"].sum().sort_values(ascending = False).head(5)

plt.figure(figsize = (10,6))

top_5_products.plot(kind = "bar",color = "skyblue" , edgecolor = "black")

plt.title("Top 5 Products by Revenue" , fontsize = 16)
plt.xlabel("Products", fontsize = 12)
plt.ylabel("Revenue (PKR)", fontsize = 12)

plt.ticklabel_format(style = "plain" , axis = "y")

plt.xticks(rotation =30)
plt.grid(axis = "y" , linestyle = "--", alpha = 0.5)

plt.show()
```

# PIE chart (What percentage of Revenue comes from which Category)

```python
category_sales = df.groupby("Category")["Total Sales"].sum()

category_sales.plot(kind = "pie" ,autopct = "%1.1f%%" ,  startangle = 90)

plt.title("Revenue share by Category")
plt.ylabel("")

plt.show()
```

```python

daily_revenue = df.groupby("Dates")["Total Sales"].sum()
daily_revenue

plt.figure(figsize=(12,6))

daily_revenue.plot( kind="line", marker="o", linewidth=2, color="red")

plt.title("Daily Revenue Trend", fontsize=16)
plt.xlabel("Days")
plt.ylabel("Revenue (PKR)")

plt.ticklabel_format(style="plain", axis="y")


plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

# Line Chart (Monthly revenue)

```python
df["Month"] = df["Dates"].dt.month
Monthly_revenue = df.groupby("Month")["Total Sales"].sum()
Monthly_revenue

plt.figure(figsize=(12,6))

Monthly_revenue.plot( kind="line", marker="o", linewidth=2, color="orange")

plt.title("Monthly Revenue Trend", fontsize=16)
plt.xlabel("Months")
plt.ylabel("Revenue (PKR)")

plt.ticklabel_format(style="plain", axis="y")


plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

# Histogram (Distribution of Customer Order Values)

```python
plt.figure(figsize=(10,6))

plt.hist(
    df["Total Sales"],
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Customer Order Values", fontsize=16)
plt.xlabel("Order Value (PKR)")
plt.ylabel("Number of Orders")

plt.ticklabel_format(style="plain", axis="x")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

# Box plot of Total Sales

```python
plt.figure(figsize = (8,6))

plt.boxplot(df["Total Sales"])

plt.title("Box Plot of Total Sales", fontsize = 16)
plt.ylabel("Revenue (PKR)")

plt.ticklabel_format( style ="plain" , axis ="y")


plt.show()
```
